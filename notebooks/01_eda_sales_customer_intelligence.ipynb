# Enterprise Sales & Customer Intelligence - Exploratory Data Analysis (EDA)

This notebook conducts comprehensive exploratory data analysis across 10,000+ customer records and 30,000+ orders spanning multi-year retail transactions.


```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from pathlib import Path

sns.set_theme(style="whitegrid", palette="deep")
plt.rcParams["figure.figsize"] = (12, 6)
```

## 1. Load Power BI-Ready Processed Dataset


```python
data_path = Path('../data/processed/powerbi_sales_intelligence.csv')
if not data_path.exists():
    data_path = Path('../data/raw/orders.csv')

df = pd.read_csv(data_path)
print(f"Dataset Shape: {df.shape}")
df.head()
```

## 2. Dataset Schema and Statistical Summary


```python
print(df.info())
df.describe()
```

## 3. Monthly Revenue Trend & Seasonality


```python
if 'order_date' in df.columns:
    df['order_date'] = pd.to_datetime(df['order_date'])
    monthly = df.set_index('order_date').resample('M')['total_amount'].sum()
    
    plt.figure(figsize=(14, 5))
    plt.plot(monthly.index, monthly.values, marker='o', color='#2563eb', linewidth=2.5)
    plt.title('Monthly Gross Revenue Trajectory (2023 - 2026)', fontsize=14, fontweight='bold')
    plt.xlabel('Date')
    plt.ylabel('Revenue ($)')
    plt.show()
```

## 4. Profit Margin Distribution by Category


```python
if 'category_name' in df.columns and 'profit_margin_pct' in df.columns:
    plt.figure(figsize=(12, 6))
    sns.boxplot(data=df, x='category_name', y='profit_margin_pct', palette='Set2')
    plt.title('Profit Margin Distribution Across Product Categories', fontsize=14, fontweight='bold')
    plt.xticks(rotation=15)
    plt.ylabel('Profit Margin (%)')
    plt.show()
```

## 5. Regional Sales Contribution


```python
if 'region_name' in df.columns:
    reg_sales = df.groupby('region_name')['total_amount'].sum().sort_values(ascending=False)
    plt.figure(figsize=(8, 8))
    plt.pie(reg_sales.values, labels=reg_sales.index, autopct='%1.1f%%', colors=['#2563eb', '#06b6d4', '#10b981', '#f59e0b'])
    plt.title('Market Share Contribution by Geographic Region', fontsize=14, fontweight='bold')
    plt.show()
```
